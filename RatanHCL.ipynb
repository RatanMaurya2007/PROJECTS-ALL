```python
import pandas as pd
df = pd.read_csv('/content/SuperMarket Analysis.csv')
df.head()
```

```python
# Count transactions by branch and product line
transaction_counts = df.groupby(['Branch', 'Product line']).size().reset_index(name='Transaction Count')
display(transaction_counts)
```

```python
# Calculate overall total and average sales
total_sales = df['Sales'].sum()
average_sales = df['Sales'].mean()

print(f"Overall Total Sales: ${total_sales:,.2f}")
print(f"Overall Average Sales per Transaction: ${average_sales:,.2f}")
```

```python
# Calculate total and average sales broken down by Branch and Product line
sales_metrics = df.groupby(['Branch', 'Product line'])['Sales'].agg(['sum', 'mean']).rename(columns={'sum': 'Total Sales', 'mean': 'Average Sales'}).reset_index()
display(sales_metrics)
```

```python
# Calculate the average sales for each product line and sort them in descending order
avg_sales_by_product = df.groupby('Product line')['Sales'].mean().reset_index(name='Average Sales')
avg_sales_by_product = avg_sales_by_product.sort_values(by='Average Sales', ascending=False)

display(avg_sales_by_product)

highest_product_line = avg_sales_by_product.iloc[0]
print(f"\nThe product line with the highest average sales is '{highest_product_line['Product line']}' with an average of ${highest_product_line['Average Sales']:.2f}.")
```

```python
# Count transactions by payment method and customer type
payment_customer_counts = df.groupby(['Payment', 'Customer type']).size().reset_index(name='Transaction Count')
display(payment_customer_counts)
```

```python
# Calculate the average customer rating for each branch
avg_rating_by_branch = df.groupby('Branch')['Rating'].mean().reset_index(name='Average Rating')
avg_rating_by_branch = avg_rating_by_branch.sort_values(by='Average Rating', ascending=False)

display(avg_rating_by_branch)
```

```python
# Find and display the 10 highest-value transactions
top_10_transactions = df.sort_values(by='Sales', ascending=False).head(10)
display(top_10_transactions)
```

```python
# Check for missing values in each column
missing_values = df.isnull().sum().reset_index(name='Missing Values Count')
print("--- Missing Values Count Per Column ---")
display(missing_values)

# Check for duplicate rows
duplicate_count = df.duplicated().sum()
print(f"\nTotal duplicate rows in dataset: {duplicate_count}")
```
